```python
from statistics import correlation

import pandas as pd
import numpy as np

df = pd.read_csv('data/df_iv.csv')
print(df)
```

```python
import statsmodels.formula.api as smf

model = smf.ols('y~is_assigned', data = df).fit()
print(model.summary())
```

```python
model = smf.ols('is_delivered~is_assigned', data = df).fit()
print(model.summary())
```

```python
from linearmodels import IV2SLS
iv_sls = IV2SLS.from_formula('y~ 1+ [is_delivered~is_assigned]', data=df).fit(cov_type = 'unadjusted')
print(iv_sls.summary.tables[1])
```
